```python
import pandas as pd

# === 1️⃣ Load your data ===

df = pd.read_csv(r"C:/Users/Grace/EnvironmentDataScience/group_project/ca_no2_2019.csv")

# === 2️⃣ Create station_col ===
# state_code (2 digits), county_code (3 digits), site_code (4 digits)
df["station_col"] = df["Site ID"]

# === 3️⃣ Convert Date to datetime ===
df['Date'] = pd.to_datetime(df['Date'])

# === 4️⃣ Column for NO2 value ===
no2_col = "Daily Max 1-hour NO2 Concentration"  # If your NO₂ column is this one

# === 5️⃣ Group by station + date and take the average NO₂ ===
df_avg = (
    df.groupby(['station_col', 'Date'], as_index=False)
      .agg({no2_col: 'mean'})
)

# === 6️⃣ Keep the first non-NO₂ record per group ===
other_cols = [
    col for col in df.columns 
    if col not in ['station_col', 'Date', no2_col]
]

df_first = (
    df.sort_values(['station_col', 'Date'])
      .drop_duplicates(['station_col', 'Date'], keep='first')[['station_col', 'Date'] + other_cols]
)

# === 7️⃣ Merge average NO₂ back with the remaining info ===
df_final = df_avg.merge(df_first, on=['station_col', 'Date'], how='left')


# === 8️⃣ Save output (optional) ===
df_final.to_csv(r"C:/Users/Grace/EnvironmentDataScience/group_project/ca_no2_2019_avg.csv", index=False)

print("Done! Final cleaned dataset shape:", df_final.shape)
df_final.head()
```

```python
# Set the California NO₂ 1-hour standard
ca_standard = 30  # ppb (Parts Per Billion)

# Calculate the number of days exceeding the California standard for each site
def count_exceedances(group, threshold):
    """
    Counts the number of days where the 'Daily Max 1-hour NO2 Concentration'
    in a site group exceeds the specified threshold (30 ppb).
    """
    # Returns the count of days where the concentration is greater than the 30 ppb threshold.
    return (group['Daily Max 1-hour NO2 Concentration'] > threshold).sum()

# Group by site and count exceedances using the California standard
# 'filter_no2_2020' is the DataFrame containing the monitoring data.
exceed_ca = filter_no2_2020.groupby('Local Site Name').apply(
    count_exceedances,
    threshold=ca_standard,  # Using the stricter 30 ppb California standard
    include_groups=False
).reset_index(name='Days_Exceed_CA') # Rename the result column

# Merge Site Longitude and Latitude information
# Uses the first recorded coordinate pair for each unique site name.
site_coords = filter_no2_2020.groupby('Local Site Name')[['Site Longitude', 'Site Latitude']].first().reset_index()

# Merge the exceedance data with the coordinate data
exceed_data = site_coords.merge(exceed_ca, on='Local Site Name')

# Select the metric for visualization (days exceeding CA standard)
exceed_data['Exceed_Days'] = exceed_data['Days_Exceed_CA']
print(exceed_data['Exceed_Days'].head(100).to_string())
print(exceed_data.iloc[54])

# Calculate and print summary statistics
max_days = exceed_data['Exceed_Days'].max()
print(f"Maximum Exceedance Days (CA 30 ppb): {max_days} days")
print(f"Number of Monitoring Sites: {len(exceed_data)}")


# --- Data Visualization (Mapping) ---
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# 1. Create the figure and GeoAxes (Map)
fig = plt.figure(figsize=(12, 9))
ax = plt.axes(projection=ccrs.PlateCarree()) # Standard latitude/longitude projection
# Set map boundaries to focus on California
ax.set_extent([-126, -113, 30, 43], crs=ccrs.PlateCarree())

# 2. Add Geographical Features
ax.add_feature(cfeature.COASTLINE.with_scale('10m'), edgecolor='black', linewidth=0.8) # Coastline
ax.add_feature(cfeature.STATES.with_scale('10m'), edgecolor='gray', linewidth=0.5)    # State borders
ax.add_feature(cfeature.LAKES.with_scale('10m'), edgecolor='blue', facecolor='none', alpha=0.5) # Lake outlines


# 3. Plot the Scatter Plot: Color represents exceedance days
sc = ax.scatter(exceed_data['Site Longitude'], exceed_data['Site Latitude'],
                c=exceed_data['Exceed_Days'],     # Color map variable
                cmap='Reds',                      # Color gradient (darker red = more exceedance days)
                s=50,                             # Marker size
                edgecolors='k',                   # Black outline
                linewidth=0.5,
                transform=ccrs.PlateCarree(),     # Coordinate system for plotting data
                vmin=0, vmax=max_days,            # Color normalization range
                zorder=10)

# 4. Add Colorbar
cbar = plt.colorbar(sc, ax=ax, shrink=0.6, pad=0.05)
# Update label to reflect the 30 ppb California standard
cbar.set_label('Days Exceeding California 1-hr NO₂ Standard (30 ppb) in 2020', fontsize=12)

# 5. Title and Aesthetics
ax.set_title('California NO₂ Monitoring Sites: Days Exceeding California Standard (2020)\n'
             'Red dots = higher exceedance days (Standard: 30 ppb)',
             fontsize=14, pad=20)

# Optional: Add Gridlines
gl = ax.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.5, linestyle='--')
gl.top_labels = False    # Disable top labels
gl.right_labels = False  # Disable right labels

# Adjust layout and display
plt.tight_layout()
plt.show()
```
